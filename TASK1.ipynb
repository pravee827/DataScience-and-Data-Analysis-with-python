```python
# Student Performance Analysis using Python
# ==========================================
# Student Performance Data Analysis Project
# ==========================================

# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

# Ignore warning messages
warnings.filterwarnings('ignore')

# ------------------------------------------
# Step 1: Load Dataset
# ------------------------------------------

# Replace the file name with your dataset path if needed
df = pd.read_csv('student-mat.csv',sep=';')

print('Dataset Loaded Successfully!')

# ------------------------------------------
# Step 2: Explore Dataset
# ------------------------------------------

print('\nFirst 5 Rows:')
print(df.head())

print('\nDataset Shape:')
print(df.shape)

print('\nColumn Data Types:')
print(df.dtypes)

print('\nMissing Values:')
print(df.isnull().sum())

# ------------------------------------------
# Step 3: Remove Duplicates
# ------------------------------------------

before_duplicates = df.shape[0]

df = df.drop_duplicates()

after_duplicates = df.shape[0]

print('\nDuplicate Rows Removed:', before_duplicates - after_duplicates)

# ------------------------------------------
# Step 4: Data Analysis
# ------------------------------------------

# Average Final Grade
average_grade = df['G3'].mean()
print('\nAverage Final Grade (G3):', round(average_grade, 2))

# Students scoring above 15
students_above_15 = df[df['G3'] > 15].shape[0]
print('Number of students scoring above 15:', students_above_15)

# Correlation between study time and final grade
correlation = df['studytime'].corr(df['G3'])
print('Correlation between Study Time and Final Grade:', round(correlation, 2))

# Average score by gender
average_by_gender = df.groupby('sex')['G3'].mean()
print('\nAverage Score by Gender:')
print(average_by_gender)

# ------------------------------------------
# Step 5: Data Visualization
# ------------------------------------------

# Set plot style
sns.set(style='whitegrid')

# Histogram of Final Grades
plt.figure(figsize=(8, 5))
plt.hist(df['G3'], bins=10)
plt.title('Distribution of Final Grades')
plt.xlabel('Final Grade (G3)')
plt.ylabel('Number of Students')
plt.show()

# Scatter Plot: Study Time vs Final Grade
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df['studytime'], y=df['G3'])
plt.title('Study Time vs Final Grade')
plt.xlabel('Study Time')
plt.ylabel('Final Grade (G3)')
plt.show()

# Bar Chart: Male vs Female Average Score
plt.figure(figsize=(6, 5))
average_by_gender.plot(kind='bar')
plt.title('Average Final Grade by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Final Grade (G3)')
plt.xticks(rotation=0)
plt.show()

# ------------------------------------------
# Step 6: Conclusion
# ------------------------------------------

print('\nConclusion:')
print('- The dataset was cleaned successfully.')
print('- Average final grades were calculated.')
print('- Relationship between study time and grades was analyzed.')
print('- Visualizations were created successfully.')
```
